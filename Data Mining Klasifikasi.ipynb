Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

# Misalnya df sudah didefinisikan sebelumnya dan kolom 'Profit' ada

# Membuat label berdasarkan Profit
df['Label'] = df['Profit'].apply(lambda x: 1 if x > 0 else 0)

# Menyiapkan fitur dan label
features = df[['Sales', 'Discount', 'Quantity']].fillna(0)
labels = df['Label']

# Split data menjadi train dan test
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.3, random_state=42)

# Membuat dan melatih model Decision Tree
model = DecisionTreeClassifier()
model.fit(X_train, y_train)

# Memprediksi hasil
y_pred = model.predict(X_test)

# Menampilkan hasil evaluasi
print("Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import pandas as pd

# Membaca data
df = pd.read_excel("Datasets Superstore.xlsx")

# Menambahkan kolom label: 1 jika profit > 0, else 6
df["Label"] = df["Profit"].apply(lambda x: 1 if x > 0 else 6)

# Menyiapkan fitur dan label
features = df[['Sales', 'Discount', 'Quantity']].fillna(0)
labels = df['Label']

# Membagi data latih dan uji
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.3, random_state=42)

# Inisialisasi dan pelatihan model
model = KNeighborsClassifier()
model.fit(X_train, y_train)

# Prediksi
y_pred = model.predict(X_test)

# Evaluasi model
print("KNN (K-Nearest Neighbors)")
print("Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import pandas as pd

# Baca data
df = pd.read_excel("Datasets Superstore.xlsx")

# Buat label: 1 jika profit > 0, selain itu 6
df["Label"] = df["Profit"].apply(lambda x: 1 if x > 0 else 6)

# Siapkan fitur dan label
features = df[['Sales', 'Discount', 'Quantity']].fillna(0)
labels = df['Label']

# Split data
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.3, random_state=42)

# Model Naive Bayes
model = GaussianNB()
model.fit(X_train, y_train)

# Prediksi
y_pred = model.predict(X_test)

# Evaluasi
print("Naive Bayes (GaussianNB)")
print("Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
